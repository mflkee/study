- Для категориальных признаков (`Пол`, `Кредитная карта`) относительно целевой переменной целесообразно использовать критерий хи-квадрат.
- Если выявлены сильные нелинейные связи, можно дополнительно использовать корреляцию Спирмена или Кендалла.


## Задание 3. Исследование влияния категориальных признаков на целевой


```python
cat_features = ['Пол', 'Кредитная карта']

for col in cat_features:
    ct = pd.crosstab(df[col], df['Статус сотрудничества'])
    print(f'Кросс-таблица {col} vs Статус сотрудничества:')
    print(ct)
    print()
    print('Проценты по строкам:')
    print((ct.div(ct.sum(axis=1), axis=0) * 100).round(2))
    print()
    chi2, p, dof, expected = stats.chi2_contingency(ct)
    print(f'Хи-квадрат = {chi2:.4f}, p-value = {p:.4e}')
    print()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for i, col in enumerate(cat_features):
    ct_pct = pd.crosstab(df[col], df['Статус сотрудничества'], normalize='index') * 100
    ct_pct.plot(kind='bar', ax=axes[i], rot=0)
    axes[i].set_title(f'{col} vs Статус сотрудничества')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Доля, %')
    axes[i].legend(title='Статус сотрудничества')

plt.tight_layout()
plt.show()
```

### Комментарии и выводы по заданию 3

**3.1)** Предположения о наличии зависимости:

- `Пол`: женщины уходят из банка чаще (55.8%), чем мужчины (41.6%). Разница статистически значима (χ² = 16.38, p-value ≈ 0.0001 < 0.05). Следовательно, пол влияет на вероятность ухода.
- `Кредитная карта`: доля уходов среди клиентов с кредитной картой (49.8%) и без нее (43.9%) различается незначительно. Различие статистически незначимо (χ² = 1.59, p-value ≈ 0.21 > 0.05). Следовательно, наличие кредитной карты, по-видимому, не оказывает существенного влияния на статус сотрудничества в данной выборке.

**3.2)** Выводы по анализу зависимостей:

- Категориальный признак `Пол` является значимым предиктором ухода клиента и должен быть включен в модель.
- Признак `Кредитная карта` не показал статистически значимой связи с целевой переменной, однако его включение в модель логистической регрессии оправдано, поскольку после совместного учета с другими признаками его вклад может измениться.
- В целом, категориальные признаки имеют разную прогнозную ценность, что важно учитывать при интерпретации модели.


## Задание 4. Подготовка данных для построения модели логистической регрессии


```python
# Определяем признаки и целевую переменную
feature_cols = ['Возраст', 'Пол', 'Срок сотрудничества', 'Кредитный рейтинг',
                'Сумма на банковском счете', 'Количество банковских продуктов', 'Кредитная карта']

X = df[feature_cols].copy()
y = (df['Статус сотрудничества'] == 'Покинул банк').astype(int)

num_cols = ['Возраст', 'Срок сотрудничества', 'Кредитный рейтинг', 'Сумма на банковском счете', 'Количество банковских продуктов']
cat_cols = ['Пол', 'Кредитная карта']

print('Числовые признаки:', num_cols)
print('Категориальные признаки:', cat_cols)
print('Целевая переменная: 1 — Покинул банк, 0 — Остается клиентом')
print('Распределение целевой переменной:')
print(y.value_counts())
```

```python
# Разделение на обучающую и тестовую выборки
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42, stratify=y
)

print(f'Размер обучающей выборки: {X_train.shape[0]}')
print(f'Размер тестовой выборки: {X_test.shape[0]}')
print(f'Доля уходов в обучающей выборке: {y_train.mean():.4f}')
print(f'Доля уходов в тестовой выборке: {y_test.mean():.4f}')
```

```python
# Пайплайн предобработки
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(drop='first', sparse_output=False), cat_cols)
    ]
)

print('Предобработка:')
print('- Числовые признаки масштабируются StandardScaler.')
print('- Категориальные признаки кодируются OneHotEncoder с drop="first" для исключения мультиколлинеарности.')
print('- Обучение кодировщиков и масштабировщика производится только на train, к test применяется transform.')
```

### Комментарии и выводы по заданию 4

**4.1)** Обоснование выбранных методов предварительной обработки:

- **Исключение `ID клиента`**: уникальный идентификатор не несет обобщающей информации и может привести к переобучению, поэтому удален из признаков.
- **Стратифицированное разбиение (`train_test_split` с `stratify=y`)**: сохраняет пропорции классов в обучающей и тестовой выборках, что важно для корректной оценки качества при бинарной классификации.
- **StandardScaler для числовых признаков**: логистическая регрессия чувствительна к масштабу признаков. Масштабирование обеспечивает сходимость оптимизации и корректную интерпретацию коэффициентов.
- **OneHotEncoder с `drop='first'`** для бинарных категориальных признаков (`Пол`, `Кредитная карта`) преобразует категории в числовой вид, избегая ложной порядковой интерпретации и мультиколлинеарности.
- **Принцип `fit` на train, `transform` на test**: предотвращает утечку информации из тестовой выборки, обеспечивая реалистичную оценку обобщающей способности модели.


## Задание 5. Обучение модели логистической регрессии


```python
# Создаем и обучаем модель
model = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(random_state=42, max_iter=1000))
])

model.fit(X_train, y_train)
print('Модель успешно обучена.')
```

```python
# Получаем коэффициенты модели
feature_names_out = num_cols + list(
    model.named_steps['preprocessor']
    .named_transformers_['cat']
    .get_feature_names_out(cat_cols)
)

coefs = model.named_steps['classifier'].coef_[0]
intercept = model.named_steps['classifier'].intercept_[0]

coef_df = pd.DataFrame({'Признак': feature_names_out, 'Коэффициент': coefs})
coef_df = coef_df.sort_values('Коэффициент', key=abs, ascending=False)
print(coef_df.round(6))
print()
print(f'Свободный член (intercept): {intercept:.6f}')
```

```python
# Оценка на обучающей выборке
y_train_pred = model.predict(X_train)
y_train_proba = model.predict_proba(X_train)[:, 1]

print('Метрики на обучающей выборке:')
print(f'Accuracy:  {accuracy_score(y_train, y_train_pred):.4f}')
print(f'Precision: {precision_score(y_train, y_train_pred):.4f}')
print(f'Recall:    {recall_score(y_train, y_train_pred):.4f}')
print(f'F1-score:  {f1_score(y_train, y_train_pred):.4f}')
print(f'ROC-AUC:   {roc_auc_score(y_train, y_train_proba):.4f}')
print()
print('Матрица ошибок (обучение):')
print(confusion_matrix(y_train, y_train_pred))
```

### Комментарии и выводы по заданию 5

**5.1)** Гиперпараметры модели:

- `LogisticRegression` с параметрами по умолчанию (`C=1.0`, `penalty='l2'`, `solver='lbfgs'`) и `max_iter=1000`.
- `random_state=42` для воспроизводимости результатов.
- Выбор обоснован тем, что это базовая линейная модель, хорошо подходящая для бинарной классификации с интерпретируемыми коэффициентами. Регуляризация L2 по умолчанию помогает контролировать переобучение. Увеличение `max_iter` гарантирует сходимость оптимизации.

**5.2)** Признаки переобучения:

- Метрики на обучающей выборке (accuracy ≈ 0.70, ROC-AUC ≈ 0.75) не сильно выше ожидаемых метрик на тесте (см. задание 6), поэтому явного переобучения не наблюдается. Модель обобщается на новые данные.

**5.3)** Уравнение модели прогнозирования:

После обучения логистической регрессии линейная комбинация признаков имеет вид:

z = w0 + w1*x_Возраст_стд + w2*x_Срок_стд + w3*x_КредРейтинг_стд + w4*x_Сумма_стд + w5*x_Продукты_стд + w6*x_Пол_Мужской + w7*x_КредитнаяКарта_Нет

где x_стд — стандартизированные значения числовых признаков, а бинарные признаки равны 0 или 1.

Вероятность ухода клиента:

P(Покинул) = 1 / (1 + exp(-z))

Конкретные значения коэффициентов приведены в таблице выше. Наибольший положительный вклад в увеличение вероятности ухода вносит `Возраст`, наибольший отрицательный — `Пол_Мужской` (мужчины уходят реже).

**5.4)** Интерпретация метрик на обучающей выборке:

- **Accuracy ≈ 0.70**: модель правильно классифицирует около 70% клиентов.
- **Precision ≈ 0.71**: из предсказанных уходов около 71% действительно уходят.
- **Recall ≈ 0.64**: модель находит около 64% всех реальных уходов.
- **F1-score ≈ 0.67**: баланс между точностью и полнотой.
- **ROC-AUC ≈ 0.75**: модель обладает удовлетворительной разделительной способностью.


## Задание 6. Оценка качества модели на тестовой выборке


```python
# Предсказания на тестовой выборке
y_test_pred = model.predict(X_test)
y_test_proba = model.predict_proba(X_test)[:, 1]

print('Метрики на тестовой выборке:')
print(f'Accuracy:  {accuracy_score(y_test, y_test_pred):.4f}')
print(f'Precision: {precision_score(y_test, y_test_pred):.4f}')
print(f'Recall:    {recall_score(y_test, y_test_pred):.4f}')
print(f'F1-score:  {f1_score(y_test, y_test_pred):.4f}')
print(f'ROC-AUC:   {roc_auc_score(y_test, y_test_proba):.4f}')
print()
print('Матрица ошибок (тест):')
print(confusion_matrix(y_test, y_test_pred))
print()
print('Classification report (тест):')
print(classification_report(y_test, y_test_pred, target_names=['Остается', 'Покинул']))
```

```python
# ROC-кривая
fpr, tpr, thresholds = roc_curve(y_test, y_test_proba)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f'ROC curve (AUC = {roc_auc_score(y_test, y_test_proba):.4f})')
plt.plot([0, 1], [0, 1], 'k--', label='Random classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC-кривая на тестовой выборке')
plt.legend()
plt.show()
```

## Задание 7. Итоговая оценка качества прогнозирования

**7.1)** Полученная модель логистической регрессии демонстрирует на тестовой выборке следующие результаты:

- **Accuracy ≈ 0.65** — модель верно классифицирует около 65% клиентов.
- **Precision ≈ 0.65** — из клиентов, которых модель отнесла к уходящим, примерно 65% действительно уходят.
- **Recall ≈ 0.59** — модель выявляет около 59% всех уходящих клиентов.
- **F1-score ≈ 0.62** — сбалансированная оценка качества.
- **ROC-AUC ≈ 0.72** — модель имеет приемлемую разделительную способность.

**7.2)** Сравнение метрик на обучающей и тестовой выборках показывает небольшое снижение качества (например, accuracy с 0.70 до 0.65, ROC-AUC с 0.75 до 0.72), что свидетельствует об отсутствии сильного переобучения. Модель обобщается на новые данные, хотя и с умеренным качеством.

**7.3)** Наиболее значимые признаки, влияющие на вероятность ухода клиента:

- **Возраст** — более старшие клиенты склонны уходить чаще.
- **Пол** — мужчины уходят реже, чем женщины.
- **Сумма на банковском счете** — клиенты с более высокими остатками на счете чаще уходят (возможно, они более мобильны и требовательны к условиям).
- **Количество банковских продуктов**, **срок сотрудничества**, **кредитный рейтинг** и **кредитная карта** имеют меньший вклад.

**7.4)** Возможные пути улучшения модели:

1. Создание новых признаков, например категорий по возрастным группам или уровню баланса.
2. Подбор гиперпараметров логистической регрессии (например, параметра регуляризации `C`) с помощью кросс-валидации.
3. Использование других моделей (деревья решений, случайный лес, градиентный бустинг) для повышения точности.
4. Анализ порогов классификации: в задаче удержания клиентов важнее полнота (recall), поэтому можно снизить порог для выявления большей доли уходящих.

**Общий вывод:** модель логистической регрессии показывает умеренное качество прогнозирования оттока клиентов (ROC-AUC ≈ 0.72). Она может использоваться как базовый инструмент для выявления клиентов, склонных к уходу, но для практического применения рекомендуется доработка с добавлением новых признаков и подбором гиперпараметров.
